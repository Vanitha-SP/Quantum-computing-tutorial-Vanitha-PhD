```python
from qiskit import QuantumRegister, ClassicalRegister, QuantumCircuit
import matplotlib.pyplot as plt
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

qr=QuantumRegister(1)
cr=ClassicalRegister(1)
qc=QuantumCircuit(qr, cr)
qc.h(qr[0])
qc.x(qr[0])
qc.s(qr[0])
qc.measure(qr, cr)

sim=AerSimulator()
job=sim.run(qc)
result=job.result()
output=result.get_counts()
print(output)

```

```python
plot_histogram(output)
```

```python
qc.draw('mpl')
```

# Bell circuit



```python
qr=QuantumRegister(2)
cr=ClassicalRegister(2)
qc1=QuantumCircuit(qr,cr)
qc1.h(qr[0])
qc1.cx(qr[0],qr[1])
qc1.measure(qr[0],cr[0])
qc1.measure(qr[1],cr[1])
```

```python
qc1.draw('mpl')
```

```python
sim=AerSimulator()
job=sim.run(qc1)
result=job.result()
output1=result.get_counts()
print(output)
```

```python
plot_histogram(output1)
```

#GHZ circuit

```python
qr1=QuantumRegister(3, 'q')
cr1=ClassicalRegister(3, 'c')
q_c=QuantumCircuit(qr1,cr1)
q_c.h(qr1[0])
q_c.cx(qr1[0],qr1[1])
q_c.cx(qr1[1], qr1[2])
q_c.measure(qr1[0], cr1[0])
q_c.measure(qr1[1], cr1[1])
q_c.measure(qr1[2], cr1[2])
q_c.draw('mpl')
```

```python
job=AerSimulator().run(q_c)
result=job.result()
output_2=result.get_counts()
```

```python
print(output_2)
```

```python
plot_histogram(output_2)
```
